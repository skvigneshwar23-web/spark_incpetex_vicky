```python
dbutils.widgets.text("country", "")
dbutils.widgets.text("population", "50k")
dbutils.widgets.dropdown("gender", "male",["male","female"])
dbutils.widgets.combobox("party", "dmk",["bjp","congress","dmk"])
dbutils.widgets.text("range_value","5")
int_range=int(dbutils.widgets.get("range_value"))
print(int_range)
for i in range(int_range):
  print(i)
```
