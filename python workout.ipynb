### python map filter reduce

```python
from functools import *
from itertools import *
import operator
lit1=[1,2,3,4,5,6]
a=reduce(lambda x,y:x+y,lit1)
print(a)

a=reduce(operator.sub,lit1)
print(a)

a=accumulate(lit1,lambda x,y:x+y)
print(list(a))

a=map(lambda x:x%2==0,lit1)
print(list(a))
a=filter(lambda x:x%2==0,lit1)
print(list(a))

a=map(lambda x:x*2,lit1)
print(list(a))
a=filter(lambda x:x*2,lit1)
print(list(a))

print(sorted(lit1,reverse=True))


print(lit1)
lit1.sort(reverse=True)
print(lit1)

li=[(9,2),(333,322),(344,1)]
print(sorted(li,key=lambda x:x[1]))
print(sorted(li,key=lambda x:x[1],reverse=True))
li.sort(key=lambda x:x[0] , reverse=True)
print(li)
li.sort(key=lambda x:x[1],reverse=True)


k=lambda b:b*10 if b>1000 else 0
print(k(45))

word=['abba','cba','def']
print(list(filter(lambda x:(x=="".join(reversed(x))),word)))
print("|".join(word))

print([x for x in range(10) if x%2==0])
print([x  if x%2==0 else 0 for x in range(10)])
print([[j for j in range(1)] for x in range(10)])

d={1:'a',2:'b',3:'c'}
print(2 in d)
ko=[{'name':'abc','age':93},{'name':'xyz','age':24}]
print(sorted(ko,key=lambda i:i['age']))
ab=dict(sorted(d.items(),key=lambda x:x[0],reverse=True))
print(ab)

a='1234'
o=slice(1,5,2)
print(a[o])

a=[4,2,56,87,87,1]
p=sorted(a,key=lambda a:a)
print(p)
a.sort(key=lambda a:a)
print(a)

print(set(a))

kop={1:"hi",2:{"3":"hello"}}
print(kop[2]["3"])


def genrerator():
    yield 1
    yield 2
for i in genrerator():
    print(i)
print(list(genrerator()))

a=genrerator()
print(next(a))
print(next(a))
```
