
This notebook is going do 2 activities

Once for all - Create all necessary path and table components needed for this project such as
Catalog(parameter) -> Schema(parameter) -> Volume (sourcedatalake ,bronze ,silver ,gold).
Defining all configuration parameters such as volume path, table information etc., once for all, rather than we hardcode all these in every notebook

```python
dbutils.widgets.text("catalog","Proddefault")
Catalog=dbutils.widgets.get("catalog").strip()
dbutils.widgets.text("schema","proddb")
Schema=dbutils.widgets.get("schema").strip()
```

```python
print(Catalog)
print(Schema)
```

```python
#putting database and catlog and schema 
spark.sql(f"create catalog if not exists {Catalog};")
spark.sql(f"create schema if not exists {Catalog}.{Schema};")
```

```python
# generating layers

spark.sql(f"create volume if not exists {Catalog}.{Schema}.datalake;")
spark.sql(f"""create volume if not exists {Catalog}.{Schema}.bronze;""")
spark.sql(f"""create volume if not exists {Catalog}.{Schema}.silver;""")
spark.sql(f"""create volume if not exists {Catalog}.{Schema}.gold;""")
```

```python
src=f"/Volumes/{Catalog}/{Schema}/datalake"
bronze=f"/Volumes/{Catalog}/{Schema}/bronze"
silver=f"/Volumes/{Catalog}/{Schema}/silver"
gold=f"/Volumes/{Catalog}/{Schema}/gold"

bronzedb=f"{Catalog}.{Schema}"
silverdb=f"{Catalog}.{Schema}"
golddb=f"{Catalog}.{Schema}"
print("source datalake location(datalake)",src)
print("bronze datalake location(datalake)",bronze)
print("silver database (lakehouse)",silverdb)
print("gold datalake (lakehouse)",golddb)
```
