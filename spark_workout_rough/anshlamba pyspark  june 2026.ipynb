```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
data = [("  rAm kuMAR  ", "ram.kumar@gmail.com", "987-654-3210", "Chennai")]
dfstr = spark.createDataFrame(data,["name", "email", "phone", "city"])

dfstr.select (
    col ("name"),
    initcap("name").alias("initcap"), #Capitalize first letter of each word.
    upper(col("name")).alias("upper"),
    lower(col("name")).alias("lower"),
    trim(col("name")).alias("trim"),
    ltrim(col("name")).alias("ltrim"),
    rtrim(col("name")).alias("rtrim"),
    regexp_replace("phone",'-','').alias("regex"),
    length("name").alias("length"),
    split("email","@").alias("split"),  #Split string into array.
    substring("name",0,3).alias("substring"), #Extract part of string
    substring_index("email","@",1).alias("substring_index"),
    #replace("city", "Chennai", "Madras").alias("replace"),
    translate("name","rR","12").alias("translate"), #Replace characters one-to-one
    regexp_replace("name","r|R","1").alias("regex_repl"), #Replace text using regex patterns
    col("email").contains("@gmail.com").alias("contains"),
    col("email").endswith("@gmail.com").alias("endswith"),
    col("email").startswith("ram").alias("startswith"),
    array_contains(split("email","@"),"gmail.com").alias("array_contains"), #Checks whether value exists inside an array column
    instr("email","@").alias("instr"), #Find position of a substring.
    reverse("name").alias("reverse"), #Reverses a string.
    repeat("city",2).alias("repeat"),
    concat(col("city"), lit("_India")).alias("concat"),
    concat(col("city"), col("phone")).alias("concatcolumns"), #Joins columns together.
    concat_ws("-", col("city"), lit("TamilNadu")).alias("concat_ws") #ws = with separator
).display(truncate=False)
```

# group by 

groupBy() is used to group records based on one or more columns and perform aggregate operations such as count, sum, avg, min, and max. It is similar to SQL's GROUP BY clause and is widely used for reporting, analytics, and data summarization in Spark.


Common aggregation functions in PySpark include count(), sum(), avg(), mean(), min(), max(), first(), last(), countDistinct(), collect_list(), collect_set(), stddev(), variance(), and approx_count_distinct().

```python
#df.printSchema()
#scenario1
from pyspark.sql.functions import sum as _sum
df.groupby('Item_Type').agg(sum('Item_MRP')).show()
df.groupby('Item_Type').agg(avg('Item_MRP')).show()

#scenario2 
df.groupby('Item_Type','Outlet_size').agg(avg('Item_MRP')).show()

#scenario3
df.groupby('Item_Type','Outlet_size').agg(avg('Item_MRP'),sum('Item_MRP')).show()
```

```python
from pyspark.sql.functions import *

data = [
    ("IT", 50000),
    ("IT", 60000),
    ("HR", 40000),
    ("HR", 45000),
    ("Sales", 70000)
]

dfagg = spark.createDataFrame(
    data,
    ["dept", "salary"]
)

dfagg.groupBy("dept").agg(
    count("*").alias("emp_count"),
    sum("salary").alias("total_salary"),
    avg("salary").alias("avg_salary"),
    max("salary").alias("max_salary"),
    min("salary").alias("min_salary"),
    countDistinct("salary").alias("countDistinct"),
    first("salary").alias("first"),
    last("salary").alias("last"),
    stddev("salary").alias("stddev"),
    variance("salary").alias("variance"), 
    sumDistinct("salary").alias("sumDistinct")
).show()
```

# collect_list()
Aggregates values into an array and preserves duplicates.
Example:
[Ram, Sam, Ram]

# collect_set()
Aggregates values into an array and removes duplicates.
Example:
[Ram, Sam]

# approx_count_distinct()
Returns an approximate count of distinct values using a probabilistic algorithm. It is faster and more memory-efficient than countDistinct() for large datasets.

```python
from pyspark.sql.functions import *

data = [
('IT','Ram'),
('IT','Sam'),
('IT','Ram'),
('HR','John')
]
df_collect = spark.createDataFrame(
    data,
    ["dept", "name"]
)

df_collect.groupBy("dept").agg(collect_list("name")).show()
df_collect.groupBy("dept").agg(collect_set("name")).show()
df_collect.groupBy("dept").agg(approx_count_distinct("name")).show()
```

### PIVOT

It is used to convert row values into columns.
pivot() is used to transform distinct values from a column into multiple columns and perform aggregations on them. It is commonly used to reshape data, similar to an Excel Pivot Table. It must be used with groupBy() and an aggregate function such as sum(), count(), or avg().

```python
from pyspark.sql.functions import sum

data = [
    ("IT", "Jan", 1000),
    ("IT", "Feb", 2000),
    ("HR", "Jan", 1500),
    ("HR", "Feb", 2500)
]

df_pivot = spark.createDataFrame(
    data,
    ["dept", "month", "salary"]
)

df_pivot.show()

df_pivot.groupBy("month").pivot("dept").agg(sum("salary")).display()
```

A join is used to combine data from two DataFrames based on a common column.
SYNTAX:
df1.join(df2, join_condition, join_type)

```python
#DF
emp = [
    (1, "Ram", 101),
    (2, "Sam", 102),
    (3, "John", 103),
    (4, "Ravi", 104)
]
emp_df = spark.createDataFrame(
    emp,
    ["emp_id", "emp_name", "dept_id"]
)
#DF2
dept = [
    (101, "IT"),
    (102, "HR"),
    (105, "Finance")
]

dept_df = spark.createDataFrame(
    dept,
    ["dept_id", "dept_name"]
)
```

```python
#INNER JOIN
emp_df.join(dept_df,emp_df.dept_id==dept_df.dept_id,"inner").display()
#or
emp_df.join(dept_df,emp_df['dept_id']==dept_df['dept_id']).display()
emp_df.join(dept_df,"dept_id","left").display()
emp_df.join(dept_df,emp_df.dept_id==dept_df.dept_id,"right").display()
emp_df.join(dept_df,"dept_id","outer").display()
emp_df.join(dept_df,"dept_id","left_semi").display()  #Returns only left DataFrame columns.
emp_df.join(dept_df,"dept_id","left_anti").display()  #Returns only unmatched left rows.
emp_df.crossJoin(dept_df).display()  #Returns the Cartesian product.
emp_df.join(dept_df,"dept_id","anti").display()  #Returns only unmatched  rows
```

```python
#row_number()

from pyspark.sql.window import Window
df.withColumn('row_cal',row_number().over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').desc()))).display()

#rank()

from pyspark.sql.window import Window
df.withColumn('rank_cal',rank().over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#dense_rank()

from pyspark.sql.window import Window
df.withColumn('dense_rank_cal',dense_rank().over(Window.orderBy(col('Item_MRP').desc()))).display()

#lead()

from pyspark.sql.window import Window
df.withColumn('lead',lead('Item_MRP',2).over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#lag()

from pyspark.sql.window import Window
df.withColumn('lag',lag('Item_MRP',2).over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#ntile()
from pyspark.sql.window import Window
df.withColumn('ntile',ntile(3).over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#cume_dist()

from pyspark.sql.window import Window
df.withColumn('cume_dist',cume_dist().over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#first()

from pyspark.sql.window import Window
df.withColumn('first',first('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#last()

from pyspark.sql.window import Window
df.withColumn('last',last('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#nth_value()

from pyspark.sql.window import Window
df.withColumn('nth_value',nth_value('Item_MRP',2).over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#percent_rank()
from pyspark.sql.window import Window
df.withColumn('percent_rank',percent_rank().over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()     


#avg()

from pyspark.sql.window import Window
df.withColumn('avg',avg('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#sum()

from pyspark.sql.window import Window
df.withColumn('sum',sum('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#count()

from pyspark.sql.window import Window
df.withColumn('count',count('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#max()

from pyspark.sql.window import Window
df.withColumn('max',max('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#min()

from pyspark.sql.window import Window
df.withColumn('min',min('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()

#stddev()

from pyspark.sql.window import Window
df.withColumn('stddev',stddev('Item_MRP').over(Window.partitionBy('Item_Type').orderBy(col('Item_MRP').asc()))).display()


```

# EXTERNAL AND MANAGED TABLE 

A managed table is fully controlled by Spark, including both metadata and data files. Dropping a managed table removes both metadata and data. An external table stores metadata in the metastore but keeps data at a user-defined location; dropping the table removes only metadata and leaves the underlying data files intact.

# Managed Table
= Spark owns data + metadata
A managed table is fully controlled by Spark/Databricks.

**CREATE** 
df.write.saveAsTable("employees")
**STORAGE** **A** 
dbfs:/user/hive/warehouse/employees

**DELETE** 
Metadata Deleted
Data Deleted

Spark
 │
 ├── Metadata
 └── Data Files

# External Table
An external table points to data stored at a specific location.
= Spark owns metadata only
**CREATE** 
df.write \
  .option("path", "/mnt/raw/employees") \
  .saveAsTable("employees_ext")

**DELETE**
Metadata Deleted
Data Remains


Spark
 │
 ├── Metadata
 └── Data Files
